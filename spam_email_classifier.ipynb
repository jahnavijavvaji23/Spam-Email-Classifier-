```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB

# Sample email dataset (1 = Spam, 0 = Not Spam / Ham)
emails = [
    ("Win a free iPhone now! Click here to claim your prize", 1),
    ("Congratulations! You have won a $1000 cash lottery reward", 1),
    ("Urgent: Your bank account is suspended. Verify details immediately", 1),
    ("Cheap meds and weight loss pills available online today", 1),
    ("Limited time offer! Earn money from home with zero investment", 1),
    ("Claim your free gift card before the offer expires tonight", 1),
    ("You have been selected for an exclusive crypto giveaway", 1),
    ("Act now! Get 90% off luxury watches at our clearance sale", 1),
    ("Hi Jahnavi, are we still meeting at 10 AM for the project review?", 0),
    ("Please find attached the agenda and notes from today's meeting", 0),
    ("Can you share the updated slides before tomorrow morning?", 0),
    ("Lunch at the usual spot around 1:30 PM today?", 0),
    ("Your electricity bill for this month has been generated", 0),
    ("Reminder: team sync call scheduled for Thursday afternoon", 0),
    ("Thanks for sending over the code earlier, I will review it today", 0),
    ("Let me know when you reach home safely after class", 0),
]

texts = [item[0] for item in emails]
labels = [item[1] for item in emails]

# Convert text into numerical TF-IDF features
vectorizer = TfidfVectorizer(stop_words="english", lowercase=True)
X = vectorizer.fit_transform(texts)

# Split into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.25, random_state=42
)

# Train Naive Bayes classifier
model = MultinomialNB()
model.fit(X_train, y_train)

# Evaluate on test split
y_pred = model.predict(X_test)
print(f"Test Accuracy: {accuracy_score(y_test, y_pred):.2f}\n")
print(classification_report(y_test, y_pred, target_names=["Not Spam", "Spam"], zero_division=0))

# Interactive test loop
def predict_email(message: str) -> str:
    vec = vectorizer.transform([message])
    pred = model.predict(vec)[0]
    return "SPAM" if pred == 1 else "NOT SPAM"

print("Type an email message to test (or type 'exit' to quit):\n")
while True:
    msg = input("Email: ").strip()
    if msg.lower() in ("exit", "quit"):
        break
    if not msg:
        continue
    print("Prediction:", predict_email(msg), "\n")
```
